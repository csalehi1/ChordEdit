```python
import shutil
import subprocess
import time
from pathlib import Path

import pandas as pd

DATA_ROOT = Path("/shared/ssd_30T/mirick/datasets/ultra_edit/UltraEdit_Region_10")
MODEL_ROOT = "/shared/ssd_30T/mirick/models/sd-turbo"
SCRIPT_DIR = Path("/data/home/mirick/ChordEdit/daniel-grid-optimizations/daniel_grid_optimizations")
MAX_SAMPLES = 10
GPU = 6

OUTPUT_BAK = SCRIPT_DIR / "generated_bak"
OUTPUT_OPT = SCRIPT_DIR / "generated_opt"

print(f"Data root:   {DATA_ROOT}")
print(f"Model root:  {MODEL_ROOT}")
print(f"Max samples: {MAX_SAMPLES}")
print(f"GPU:        {GPU}")
```

### Run original `grid_generate.py.bak`

```python
bak_t0 = time.perf_counter()

if OUTPUT_BAK.exists():
    shutil.rmtree(OUTPUT_BAK)

result_bak = subprocess.run(
    [
        "python", "grid_generate.py.bak",
        "--data-root", str(DATA_ROOT),
        "--model-root", MODEL_ROOT,
        "--output-root", str(OUTPUT_BAK),
        "--max-samples", str(MAX_SAMPLES),
        "--gpu", str(GPU),
    ],
    cwd=str(SCRIPT_DIR),
    text=True,
)
elapsed_bak = time.perf_counter() - bak_t0

print(result_bak.stderr)
if result_bak.returncode != 0:
    raise RuntimeError(f"Original run failed (exit {result_bak.returncode})")

print(f"\nElapsed: {elapsed_bak:.2f}s")
```

### Run optimized `grid_generate.py`

```python
opt_t0 = time.perf_counter()

if OUTPUT_OPT.exists():
    shutil.rmtree(OUTPUT_OPT)

result_opt = subprocess.run(
    [
        "python", "grid_generate.py",
        "--data-root", str(DATA_ROOT),
        "--model-root", MODEL_ROOT,
        "--output-root", str(OUTPUT_OPT),
        "--max-samples", str(MAX_SAMPLES),
        "--gpus", str(GPU),
    ],
    cwd=str(SCRIPT_DIR),
    text=True,
)
elapsed_opt = time.perf_counter() - opt_t0

print(result_bak.stderr)
if result_bak.returncode != 0:
    raise RuntimeError(f"Optimized run failed (exit {result_bak.returncode})")

print(f"\nElapsed: {elapsed_opt:.2f}s")
```

## Compare outputs

```python
import numpy as np
from PIL import Image

# Find all generated cell images in both output dirs and compare pixel values.
bak_cells = sorted(OUTPUT_BAK.rglob("cells/*.jpg")) + sorted(OUTPUT_BAK.rglob("cells/*.png"))
opt_cells = sorted(OUTPUT_OPT.rglob("cells/*.jpg")) + sorted(OUTPUT_OPT.rglob("cells/*.png"))

# Build lookup by relative path from output root
def rel_map(root, paths):
    return {str(p.relative_to(root)): p for p in paths}

bak_map = rel_map(OUTPUT_BAK, bak_cells)
opt_map = rel_map(OUTPUT_OPT, opt_cells)

common_keys = sorted(set(bak_map) & set(opt_map))
print(f"Original cells:  {len(bak_map)}")
print(f"Optimized cells: {len(opt_map)}")
print(f"Matched cells:   {len(common_keys)}")

max_diffs = []
mean_diffs = []
psnr_vals = []
for key in common_keys:
    img_bak = np.array(Image.open(bak_map[key]).convert("RGB"), dtype=np.float32)
    img_opt = np.array(Image.open(opt_map[key]).convert("RGB"), dtype=np.float32)
    diff = np.abs(img_bak - img_opt)
    max_diffs.append(diff.max())
    mean_diffs.append(diff.mean())
    mse = np.mean((img_bak - img_opt) ** 2)
    psnr_vals.append(10.0 * np.log10(255.0**2 / mse) if mse > 0 else float("inf"))

max_diffs = np.array(max_diffs)
mean_diffs = np.array(mean_diffs)
psnr_vals = np.array(psnr_vals)

print(f"\nPixel difference (0-255 scale):")
print(f"  max  pixel diff across all cells: {max_diffs.max():.2f}")
print(f"  mean pixel diff across all cells: {mean_diffs.mean():.4f}")
print(f"  cells with max diff > 1.0: {(max_diffs > 1.0).sum()}")

print(f"\nPSNR (dB) between original vs optimized cells:")
print(f"  min  PSNR: {psnr_vals.min():.2f}")
print(f"  mean PSNR: {psnr_vals.mean():.2f}")
print(f"  max  PSNR: {psnr_vals.max():.2f}")
```

## Timing summary

```python
speedup = elapsed_bak / elapsed_opt if elapsed_opt > 0 else float("inf")

summary = pd.DataFrame({
    "version": ["original (grid_generate.py.bak)", "optimized (grid_generate.py)"],
    "elapsed_s": [f"{elapsed_bak:.2f}", f"{elapsed_opt:.2f}"],
    "cells": [len(bak_map), len(opt_map)],
})
display(summary)
print(f"\nSpeedup: {speedup:.2f}x")
```

```python
import matplotlib.pyplot as plt

n_samples = MAX_SAMPLES
cells_per_sample = len(opt_map) // n_samples if n_samples else 0

import subprocess as _sp
_short_hash = _sp.check_output(["git", "rev-parse", "--short", "HEAD"], cwd=str(SCRIPT_DIR)).decode().strip()
labels = ["grid_generate (original)", f"grid_generate ({_short_hash})"]
times_s = [elapsed_bak, elapsed_opt]
colors = ["#1F77B4", "#55A868"]

fig, ax = plt.subplots(figsize=(6, 4))
bars = ax.bar(labels, times_s, color=colors, width=0.55)
ax.set_ylabel("Wall time (s)")
ax.set_title(f"Grid generate runtime ({n_samples} images, {cells_per_sample} cells/image)")
ax.bar_label(bars, labels=[f"{t:.1f}s" for t in times_s], padding=4)
ax.text(
    0.5, 0.92, f"{speedup:.2f}x faster",
    transform=ax.transAxes, ha="center", fontsize=10,
)
ax.set_ylim(0, max(times_s) * 1.15)
fig.tight_layout()
plt.show()
```
